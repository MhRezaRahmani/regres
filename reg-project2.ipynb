```python
members = hbsir.load_table("members_properties", years=[1396])

head_edu = members[members["Member_Number"] == 1][["ID", "Education_Level"]]
```

```python
income = hbsir.load_table("employment_income", years=[1396])

hh_income = income.groupby("ID")["Annual_Continuous_Income"].sum().reset_index()
hh_income.rename(columns={"Annual_Continuous_Income": "Household_Income"}, inplace=True)
```

```python
import pandas as pd

merged = pd.merge(hh_income, head_edu, on="ID")
print(merged.head())
```

```python
mean_income_by_edu = merged.groupby("Education_Level")["Household_Income"].mean().sort_values(ascending=False)
print(mean_income_by_edu)
```

```python
import matplotlib.pyplot as plt

mean_income_by_edu.plot(kind="barh", figsize=(8, 6))
plt.xlabel("mean income")
plt.ylabel("education")
plt.tight_layout()
plt.show()
```
